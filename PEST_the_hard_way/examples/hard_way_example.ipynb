```python
true_obs = pd.read_csv('./obsvalues.dat', sep=r'\s+', index_col=0)
```

```python
true_obs.head()
```

```python
# pandas shenanigans here --> we can use the index from the pst.observation_data section to pull values from the true_obs dataframe
#                         --> and use the "obsval" column to update the obsval column in the pst.observation_data 🤯
pst.observation_data.obsval = true_obs.loc[pst.observation_data.index, 'obsval']
```
